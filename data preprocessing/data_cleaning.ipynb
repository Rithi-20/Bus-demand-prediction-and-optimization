# Data Cleaning
In this notebook, we clean the raw data by removing duplicate records and filling missing values with standard labels.
All files are saved into the `cleaned_data/` folder.


## 1. Load Raw Datasets


```python
import os
import pandas as pd
import shutil

# Auto-navigate to project root if running inside subfolder
if os.path.basename(os.getcwd()) in ["data preprocessing", "EDA"]:
    os.chdir("..")

df_demand = pd.read_csv("raw data/hourly_demand_operations.csv", low_memory=False)
df_trips = pd.read_csv("raw data/trip_operations.csv", low_memory=False)

print("Original Demand shape:", df_demand.shape)
print("Original Trips shape:", df_trips.shape)
```

## 2. Remove Duplicate Rows


```python
# Drop duplicate rows
df_demand = df_demand.drop_duplicates()
df_trips = df_trips.drop_duplicates()

print("Cleaned Demand shape (after removing duplicates):", df_demand.shape)
print("Cleaned Trips shape (after removing duplicates):", df_trips.shape)
```

## 3. Fill Missing Categorical Values


```python
# Fill missing festival_name as 'no_festival'
df_demand["festival_name"] = df_demand["festival_name"].fillna("no_festival")

# Fill missing scenario_id as 'NO_SCENARIO'
df_demand["scenario_id"] = df_demand["scenario_id"].fillna("NO_SCENARIO")

# Check remaining missing values
print("Remaining nulls in festival_name:", df_demand["festival_name"].isnull().sum())
print("Remaining nulls in scenario_id:", df_demand["scenario_id"].isnull().sum())
```

## 4. Save to `cleaned_data/` folder


```python
os.makedirs("cleaned_data", exist_ok=True)

df_demand.to_csv("cleaned_data/hourly_demand_operations.csv", index=False)
df_trips.to_csv("cleaned_data/trip_operations.csv", index=False)

print("Cleaned data successfully saved to 'cleaned_data/' folder!")
```
