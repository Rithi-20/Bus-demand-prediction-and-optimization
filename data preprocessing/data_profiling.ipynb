# Data Profiling
In this notebook, we inspect the raw data to see its structure, memory usage, statistical summary, missing values, and duplicates.


## 1. Load Data


```python
import os
import pandas as pd
import numpy as np

# Auto-navigate to project root if running inside subfolder
if os.path.basename(os.getcwd()) in ["data preprocessing", "EDA"]:
    os.chdir("..")

# Load raw datasets
df_demand = pd.read_csv("raw data/hourly_demand_operations.csv", low_memory=False)
df_trips = pd.read_csv("raw data/trip_operations.csv", low_memory=False)

print("Demand data shape:", df_demand.shape)
print("Trips data shape:", df_trips.shape)
```

## 2. Check Memory Usage


```python
# Check memory usage of datasets
demand_mem = df_demand.memory_usage(deep=True).sum() / (1024 * 1024)
trips_mem = df_trips.memory_usage(deep=True).sum() / (1024 * 1024)

print(f"Hourly Demand Data Memory Usage: {demand_mem:.2f} MB")
print(f"Trip Operations Data Memory Usage: {trips_mem:.2f} MB")
```

## 3. Hourly Demand Data Info & Statistical Summary


```python
# Check columns and data types
df_demand.info()
```

```python
# Statistical summary of numerical columns
df_demand.describe()
```

## 4. Check Missing Values and Duplicates in Hourly Demand


```python
# Missing values count per column
print("Missing values in Hourly Demand:")
print(df_demand.isnull().sum()[df_demand.isnull().sum() > 0])

# Duplicate rows count
print("\nDuplicate rows count:", df_demand.duplicated().sum())
```

## 5. Trip Operations Data Info & Statistical Summary


```python
# Trip data info
df_trips.info()
```

```python
# Trip statistical summary
df_trips.describe()
```

```python
# Check missing values and duplicates in trips
print("Missing values in Trips:")
print(df_trips.isnull().sum())

print("\nDuplicate rows count in Trips:", df_trips.duplicated().sum())
```
